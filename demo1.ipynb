```r
# Install the 'readxl' package if you haven't already
install.packages("readxl")
install.packages("sjPlot")
```

GET THE DATA

```r
# Load the 'readxl' library
library(readxl)

#Store url of data which is stored on GIT hub
excel_url <- 'https://github.com/Funtamendals-Charles/demo1/raw/refs/heads/main/FSI-2023-DOWNLOAD.xlsx'

# Download the file temporarily (readxl needs a local path or connection)
temp_file <- tempfile(fileext = ".xlsx")
download.file(excel_url, temp_file, mode = "wb")

# Read the Excel file into a data frame
fragility23 <- read_excel(temp_file)

# Clean up the temporary file
unlink(temp_file)
```

EXPLORATORY COMMANDS

```r
#READ general information about the dataframe and print it
cat("Column Info:\n")
str(fragility23)
cat("Read the first 10 rows:\n")
head(fragility23, 10)
cat("Read the last 10 rows:\n")
tail(fragility23, 10)
```

TRANSFORMATIVE COMMANDS

```r
#Get subset columns of the dataframe and store in frag23_sub
#Select only the columns matching the string vector
frag23_sub <- fragility23[, c("Country","Total","S1: Demographic Pressures", "P1: State Legitimacy", "E2: Economic Inequality")]
frag23_sub
```

```r
#Rename columns of frag23_sub
#The names attribute of the dataframe at the targeted columns are replaced by the new string vector
names(frag23_sub)[names(frag23_sub) %in% c("S1: Demographic Pressures","P1: State Legitimacy","E2: Economic Inequality")] <- c("S1", "P1", "E2")
frag23_sub
```

```r
#Get top ten countries with largest E2 measure
#First orders each row by the E2 measure and then we select only the first 10 rows as well as the Country and E2 columns
frag23_sub[order(frag23_sub$E2, decreasing = TRUE),][1:10,c("Country", "E2")]
#Get bottom ten countries with smallest E2 measure
#First orders each row by the E2 measure with the smallest values first and then we select only the first 10 rows as well as the Country and E2 columns
frag23_sub[order(frag23_sub$E2, decreasing = FALSE),][1:10,c("Country", "E2")]
```

SOME COMPUTATIONS

```r
#Get statstical description using summary function - returns quartiles of each column
summary(frag23_sub)
```

```r
#Get worst quartile in total
#Use quantile function on the total column, then select to see top 75%, ask function to ignore NA values. Store in q3_worst variable for use later.
q3_worst = quantile(frag23_sub$Total, probs=0.75, na.rm = TRUE)
q3_worst
```

```r
#Show correlation between s1, e2 and p1.
#Use the cor function, as input we take the frag23_sub and select only the relavent columns
cor(x=frag23_sub[,c("S1", "P1", "E2")])
```

```r
#Regress P1 and E2 on S1
#Use lm function. Define the ind. variables P1 and E2 and the dep. variables S1, then input the data from frag23_sub
lm(S1~P1+E2, data=frag23_sub)
#Store in model variable for later
model <- lm(S1~P1+E2, data=frag23_sub)
```

SOME PLOTTING

```r
#Give a plot for the p1 variable
#use hist function to plot a histogram, giving the P1 column as input
hist(frag23_sub$P1)
```

```r
#Visual correlation between S1 and E2
#Use plot tool to make scatter plot, specificy which columns will be on the x-axis and which will be the on y-axis
plot(x=frag23_sub$S1, y=frag23_sub$E2)
#Additionally color the points that are in the worst quartile of total
#Add the worstQt column which is a boolean vector on whether the country is in the worst quartile for the total column
frag23_sub$worstQt = frag23_sub$Total>q3_worst
#Again plots S1 on x-axis and E2 on y-axis. pch=20 changes the symbol to small points.
#Our boolean vector frag23_sub$worstQt is turned into a factor which is then passed as input to col which will assign a color mapping to the True and False values
plot(frag23_sub$S1, frag23_sub$E2, pch=20, col=as.factor(frag23_sub$worstQt))
```

```r
#Visualization of the regression P1 and E2 on S1
library(sjPlot)
plot_models(model)
```
